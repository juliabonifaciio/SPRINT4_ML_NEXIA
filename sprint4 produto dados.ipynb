## 📦 Seção 2 — Geração e Carga de Dados

> **Contexto do negócio:** Trabalhamos com dados históricos de visitação em parques nacionais brasileiros. O dataset contém 3 anos de registros diários com variáveis climáticas, temporais e de infraestrutura.

````python
# ============================================================
# GERAÇÃO DO DATASET SINTÉTICO — Realista e Documentado
# ============================================================

def gerar_dataset_turismo(n_dias=1095, seed=42):
    """
    Gera dataset sintético de visitação em parques nacionais.
    
    Parâmetros simulados:
    - Sazonalidade anual (verão/férias com +60% visitas)
    - Efeito fim de semana (+40%)
    - Efeito feriados (+80%)
    - Temperatura e chuva como moderadores
    - Tendência de crescimento linear
    - Ruído realista (σ=0.12)
    """
    np.random.seed(seed)
    parques = ['Chapada dos Veadeiros', 'Serra da Canastra', 'Iguaçu', 'Tijuca', 'Lençóis Maranhenses']
    
    datas = pd.date_range(start='2022-01-01', periods=n_dias, freq='D')
    registros = []
    
    feriados_br = [
        '2022-01-01','2022-04-21','2022-06-16','2022-09-07','2022-10-12',
        '2022-11-02','2022-11-15','2022-12-25',
        '2023-01-01','2023-04-21','2023-06-08','2023-09-07','2023-10-12',
        '2023-11-02','2023-11-15','2023-12-25',
        '2024-01-01','2024-04-21','2024-05-30','2024-09-07','2024-10-12',
        '2024-11-02','2024-11-15','2024-12-25',
    ]
    feriados_set = set(feriados_br)
    
    for parque in parques:
        capacidade_base = np.random.randint(400, 1200)
        
        for i, data in enumerate(datas):
            mes         = data.month
            dia_semana  = data.dayofweek          # 0=Seg … 6=Dom
            eh_fds      = int(dia_semana >= 5)
            eh_feriado  = int(str(data.date()) in feriados_set)
            
            # Sazonalidade (verão = dez/jan/fev, férias julho)
            sazonalidade = (
                0.60 * np.sin(2 * np.pi * (mes - 1) / 12) +          # pico em julho
                0.35 * np.sin(2 * np.pi * (mes - 12) / 12 + 0.5)     # pico em jan
            )
            
            # Temperatura (°C) — mais quente no verão
            temperatura = 22 + 10 * np.sin(2 * np.pi * (mes - 1) / 12) + np.random.normal(0, 2)
            temperatura = np.clip(temperatura, 10, 40)
            
            # Precipitação (mm) — mais chuva em jan/fev/dez
            chuva_media = 8 + 12 * abs(np.sin(2 * np.pi * (mes - 1) / 12 + 0.5))
            precipitacao = max(0, np.random.exponential(chuva_media))
            
            # Tendência de crescimento
            tendencia = 1 + 0.0002 * i
            
            # Visitação base
            visitantes = (
                capacidade_base
                * (1 + sazonalidade * 0.6)
                * (1 + 0.40 * eh_fds)
                * (1 + 0.80 * eh_feriado)
                * (1 - 0.008 * max(0, precipitacao - 5))   # chuva reduz visitação
                * (1 + 0.005 * max(0, temperatura - 20))   # calor moderado aumenta
                * tendencia
                * np.random.lognormal(0, 0.12)              # ruído multiplicativo
            )
            visitantes = max(10, int(visitantes))
            
            # Nível de demanda (target classificação)
            if visitantes >= capacidade_base * 1.3:
                nivel_demanda = 'Alta'
            elif visitantes >= capacidade_base * 0.7:
                nivel_demanda = 'Média'
            else:
                nivel_demanda = 'Baixa'
            
            registros.append({
                'data':           data,
                'parque':         parque,
                'visitantes':     visitantes,
                'temperatura':    round(temperatura, 1),
                'precipitacao':   round(precipitacao, 1),
                'eh_fds':         eh_fds,
                'eh_feriado':     eh_feriado,
                'mes':            mes,
                'dia_semana':     dia_semana,
                'sazonalidade':   round(sazonalidade, 3),
                'capacidade_base':capacidade_base,
                'nivel_demanda':  nivel_demanda,
            })
    
    return pd.DataFrame(registros)


df = gerar_dataset_turismo()

print("="*55)
print("  📊 DATASET GERADO — ESTATÍSTICAS")
print("="*55)
print(f"  Registros totais  : {len(df):,}")
print(f"  Parques           : {df['parque'].nunique()}")
print(f"  Período           : {df['data'].min().date()} → {df['data'].max().date()}")
print(f"  Visitantes (min)  : {df['visitantes'].min():,}")
print(f"  Visitantes (max)  : {df['visitantes'].max():,}")
print(f"  Visitantes (média): {df['visitantes'].mean():,.0f}")
print("="*55)
print("\n  Distribuição por nível de demanda:")
print(df['nivel_demanda'].value_counts())
print("")
df.head(8)
````

## 🏗️ Seção 4 — Engenharia de Features

> Criamos features derivadas que enriquecem o modelo com contexto temporal, climático e operacional.

````python
# ============================================================
# FEATURE ENGINEERING
# ============================================================

def criar_features(df):
    """Pipeline completo de engenharia de features."""
    df = df.copy()
    
    # ── Temporais ──────────────────────────────────────────
    df['ano']           = df['data'].dt.year
    df['trimestre']     = df['data'].dt.quarter
    df['semana_ano']    = df['data'].dt.isocalendar().week.astype(int)
    df['dia_mes']       = df['data'].dt.day
    
    # Codificação cíclica do mês (preserva periodicidade)
    df['mes_sin']       = np.sin(2 * np.pi * df['mes'] / 12)
    df['mes_cos']       = np.cos(2 * np.pi * df['mes'] / 12)
    
    # Codificação cíclica do dia da semana
    df['dia_sin']       = np.sin(2 * np.pi * df['dia_semana'] / 7)
    df['dia_cos']       = np.cos(2 * np.pi * df['dia_semana'] / 7)
    
    # ── Flags de período ───────────────────────────────────
    df['eh_verao']      = df['mes'].isin([12, 1, 2]).astype(int)
    df['eh_ferias_jul'] = (df['mes'] == 7).astype(int)
    df['eh_alta_temp']  = (df['temperatura'] >= 28).astype(int)
    df['sem_chuva']     = (df['precipitacao'] == 0).astype(int)
    df['chuva_forte']   = (df['precipitacao'] > 15).astype(int)
    
    # ── Interações ─────────────────────────────────────────
    df['fds_x_verao']   = df['eh_fds'] * df['eh_verao']
    df['fds_x_feriado'] = df['eh_fds'] * df['eh_feriado']
    df['temp_sem_chuva']= df['temperatura'] * df['sem_chuva']
    
    # ── Índice de atratividade (feature sintética) ─────────
    df['atratividade']  = (
        0.35 * df['sazonalidade'] +
        0.25 * df['eh_fds'] +
        0.20 * df['eh_feriado'] +
        0.10 * (df['temperatura'] - 15) / 20 +
        0.10 * (1 - df['precipitacao'] / 50)
    ).clip(-1, 1)
    
    # ── Encode do parque ───────────────────────────────────
    df['parque_enc'] = LabelEncoder().fit_transform(df['parque'])
    
    return df


df_feat = criar_features(df)

FEATURES = [
    'mes_sin','mes_cos','dia_sin','dia_cos',
    'eh_fds','eh_feriado','eh_verao','eh_ferias_jul',
    'temperatura','precipitacao','sazonalidade',
    'eh_alta_temp','sem_chuva','chuva_forte',
    'fds_x_verao','fds_x_feriado','temp_sem_chuva',
    'atratividade','trimestre','semana_ano','parque_enc',
]

print(f"✅ Features criadas: {len(FEATURES)} variáveis")
print("\n📋 Lista de features:")
for i, f in enumerate(FEATURES, 1):
    print(f"   {i:2d}. {f}")
````

## 🤖 Seção 5 — Modelagem: Regressão (Previsão de Visitantes)

````python
# ============================================================
# PREPARAÇÃO DOS DADOS PARA REGRESSÃO
# ============================================================
X = df_feat[FEATURES]
y_reg = df_feat['visitantes']

X_train, X_test, y_train, y_test = train_test_split(
    X, y_reg, test_size=0.20, random_state=42, shuffle=True
)

print(f"Treino : {X_train.shape[0]:,} amostras")
print(f"Teste  : {X_test.shape[0]:,} amostras")
print(f"Features: {X_train.shape[1]}")
````

````python
# ============================================================
# MODELO 1 — RANDOM FOREST (Baseline)
# ============================================================
print("🌲 Treinando Random Forest...")
rf_reg = RandomForestRegressor(
    n_estimators=200, max_depth=12,
    min_samples_leaf=4, n_jobs=-1, random_state=42
)
rf_reg.fit(X_train, y_train)
y_pred_rf = rf_reg.predict(X_test)

mae_rf  = mean_absolute_error(y_test, y_pred_rf)
rmse_rf = np.sqrt(mean_squared_error(y_test, y_pred_rf))
r2_rf   = r2_score(y_test, y_pred_rf)

print(f"   MAE  : {mae_rf:.1f} visitantes")
print(f"   RMSE : {rmse_rf:.1f} visitantes")
print(f"   R²   : {r2_rf:.4f}")
````

````python
# ============================================================
# MODELO 2 — XGBOOST COM TUNING VIA OPTUNA
# ============================================================
print("⚡ Otimizando XGBoost com Optuna (50 trials)...")

def objective(trial):
    params = {
        'n_estimators':     trial.suggest_int('n_estimators', 100, 500),
        'max_depth':        trial.suggest_int('max_depth', 3, 10),
        'learning_rate':    trial.suggest_float('learning_rate', 0.01, 0.3, log=True),
        'subsample':        trial.suggest_float('subsample', 0.6, 1.0),
        'colsample_bytree': trial.suggest_float('colsample_bytree', 0.6, 1.0),
        'reg_alpha':        trial.suggest_float('reg_alpha', 1e-4, 10, log=True),
        'reg_lambda':       trial.suggest_float('reg_lambda', 1e-4, 10, log=True),
        'random_state': 42, 'n_jobs': -1,
    }
    model = xgb.XGBRegressor(**params)
    scores = cross_val_score(model, X_train, y_train, cv=3,
                             scoring='neg_mean_absolute_error', n_jobs=-1)
    return -scores.mean()

study = optuna.create_study(direction='minimize')
study.optimize(objective, n_trials=50, show_progress_bar=True)

best_params = study.best_params
best_params.update({'random_state': 42, 'n_jobs': -1})

xgb_reg = xgb.XGBRegressor(**best_params)
xgb_reg.fit(X_train, y_train)
y_pred_xgb = xgb_reg.predict(X_test)

mae_xgb  = mean_absolute_error(y_test, y_pred_xgb)
rmse_xgb = np.sqrt(mean_squared_error(y_test, y_pred_xgb))
r2_xgb   = r2_score(y_test, y_pred_xgb)

print(f"\n📊 Melhores hiperparâmetros: {best_params}")
print(f"\n   MAE  : {mae_xgb:.1f} visitantes")
print(f"   RMSE : {rmse_xgb:.1f} visitantes")
print(f"   R²   : {r2_xgb:.4f}")
````

````python
# ============================================================
# COMPARAÇÃO DOS MODELOS DE REGRESSÃO
# ============================================================
resultados = pd.DataFrame({
    'Modelo':  ['Random Forest', 'XGBoost (Optuna)'],
    'MAE':     [mae_rf,  mae_xgb],
    'RMSE':    [rmse_rf, rmse_xgb],
    'R²':      [r2_rf,   r2_xgb],
})

fig, axes = plt.subplots(1, 3, figsize=(14, 5))
fig.suptitle('Comparação de Modelos de Regressão', fontsize=14, fontweight='bold', color='#00C9A7')

metricas = [('MAE', 'lower is better'), ('RMSE', 'lower is better'), ('R²', 'higher is better')]
for ax, (met, hint) in zip(axes, metricas):
    vals = resultados[met].values
    bars = ax.bar(resultados['Modelo'], vals, color=['#4ECDC4','#FFB400'], edgecolor='none', width=0.5)
    ax.set_title(f'{met} ({hint})', color='#94d2bd', fontsize=10)
    ax.bar_label(bars, fmt='%.3f', color='#e0e0e0', fontsize=11, padding=4)
    ax.set_ylim(0, max(vals) * 1.25)
    ax.tick_params(axis='x', rotation=10)

plt.tight_layout()
plt.savefig('comparacao_modelos.png', dpi=130, bbox_inches='tight', facecolor='#0f2027')
plt.show()

vencedor = 'XGBoost' if r2_xgb >= r2_rf else 'Random Forest'
print(f"\n🏆 Modelo vencedor: {vencedor} (R² = {max(r2_xgb, r2_rf):.4f})")

# Modelo final
modelo_final = xgb_reg if r2_xgb >= r2_rf else rf_reg
y_pred_final = y_pred_xgb if r2_xgb >= r2_rf else y_pred_rf
````

````python
# ============================================================
# GRÁFICO: REAL vs PREDITO
# ============================================================
amostra = np.random.choice(len(y_test), 300, replace=False)
y_r = np.array(y_test)[amostra]
y_p = y_pred_final[amostra]

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle(f'Avaliação do Modelo Final ({vencedor})', fontsize=14, fontweight='bold', color='#00C9A7')

# Scatter
ax1.scatter(y_r, y_p, alpha=0.5, color='#00C9A7', s=25, edgecolors='none')
lim = [min(y_r.min(), y_p.min())*0.95, max(y_r.max(), y_p.max())*1.05]
ax1.plot(lim, lim, '--', color='#FFB400', linewidth=2, label='Linha perfeita')
ax1.set_xlabel('Visitantes Reais')
ax1.set_ylabel('Visitantes Previstos')
ax1.set_title(f'Real vs Previsto — R²={max(r2_rf,r2_xgb):.4f}', color='#94d2bd')
ax1.legend()

# Resíduos
residuos = y_r - y_p
ax2.hist(residuos, bins=40, color='#A78BFA', edgecolor='none', alpha=0.8)
ax2.axvline(0, color='#FFB400', linewidth=2, linestyle='--', label='Zero')
ax2.set_xlabel('Resíduo (Real − Previsto)')
ax2.set_ylabel('Frequência')
ax2.set_title('Distribuição dos Resíduos', color='#94d2bd')
ax2.legend()

plt.tight_layout()
plt.savefig('real_vs_predito.png', dpi=130, bbox_inches='tight', facecolor='#0f2027')
plt.show()
````

## 🔍 Seção 7 — Explicabilidade com SHAP

> **Por que SHAP?** O SHAP (SHapley Additive exPlanations) atribui a cada feature uma contribuição justa para a previsão individual, baseado na teoria dos jogos cooperativos. Isso torna o modelo auditável e confiável para decisões de negócio.

````python
# ============================================================
# SHAP — IMPORTÂNCIA GLOBAL (SUMMARY PLOT)
# ============================================================
print("🔍 Calculando valores SHAP (pode levar ~30s)...")

# Amostra para eficiência
idx_sample = np.random.choice(len(X_test), min(500, len(X_test)), replace=False)
X_shap = X_test.iloc[idx_sample]

explainer = shap.TreeExplainer(modelo_final)
shap_values = explainer.shap_values(X_shap)

print("✅ SHAP calculado!")

# Summary plot
plt.figure(figsize=(10, 8))
shap.summary_plot(shap_values, X_shap, feature_names=FEATURES,
                  plot_type='bar', color='#00C9A7',
                  show=False, max_display=15)
plt.title('Importância Global das Features (SHAP)',
          fontsize=14, fontweight='bold', color='#00C9A7', pad=14)
plt.tight_layout()
plt.savefig('shap_global.png', dpi=130, bbox_inches='tight')
plt.show()
````

````python
# ============================================================
# SHAP — BEESWARM PLOT (distribuição dos valores SHAP)
# ============================================================
plt.figure(figsize=(11, 9))
shap.summary_plot(shap_values, X_shap, feature_names=FEATURES,
                  show=False, max_display=15, plot_size=(11, 9))
plt.title('Distribuição dos Valores SHAP por Feature',
          fontsize=13, fontweight='bold', pad=14)
plt.tight_layout()
plt.savefig('shap_beeswarm.png', dpi=130, bbox_inches='tight')
plt.show()
print("💡 Pontos à direita = impacto positivo na previsão; à esquerda = reduz previsão.")
````

````python
# ============================================================
# SHAP — EXPLICAÇÃO LOCAL (waterfall para 1 predição)
# ============================================================
# Selecionar uma amostra com alta visitação prevista
idx_alto = np.argmax(y_pred_final)
X_amostra_local = X_test.iloc[[idx_alto % len(X_test)]]

shap_local = explainer.shap_values(X_amostra_local)
valor_base = explainer.expected_value
previsao_local = modelo_final.predict(X_amostra_local)[0]

print(f"\n🎯 Explicação local — Previsão: {previsao_local:,.0f} visitantes")
print(f"   Valor base (média): {valor_base:,.0f} visitantes\n")

# Mostrar top contribuições
contribs = pd.DataFrame({
    'Feature':      FEATURES,
    'Valor':        X_amostra_local.values[0],
    'SHAP':         shap_local[0],
}).sort_values('SHAP', key=abs, ascending=False).head(10)

fig, ax = plt.subplots(figsize=(10, 6))
cores = ['#00C9A7' if v > 0 else '#FF6B6B' for v in contribs['SHAP']]
bars = ax.barh(contribs['Feature'], contribs['SHAP'], color=cores, edgecolor='none', height=0.6)
ax.axvline(0, color='#FFB400', linewidth=1.5, linestyle='--')
ax.set_title(f'Explicação Local — Previsão: {previsao_local:,.0f} visitantes\n'
             f'(Base: {valor_base:,.0f} | Diferença: {previsao_local-valor_base:+,.0f})',
             fontsize=12, fontweight='bold', color='#00C9A7', pad=12)
ax.set_xlabel('Contribuição SHAP (visitantes adicionais/removidos)')
ax.bar_label(bars, fmt='{:+.1f}', color='#e0e0e0', fontsize=10, padding=4)
plt.tight_layout()
plt.savefig('shap_local.png', dpi=130, bbox_inches='tight', facecolor='#0f2027')
plt.show()

# Gerar texto explicativo em linguagem natural
top3 = contribs.head(3)
motivos = []
for _, row in top3.iterrows():
    sinal = 'aumenta' if row['SHAP'] > 0 else 'reduz'
    motivos.append(f"'{row['Feature']}' {sinal} a previsão em {abs(row['SHAP']):.0f} visitantes")

print("\n🗣️  Explicação em linguagem natural:")
print(f"   Previsão de {previsao_local:,.0f} visitantes porque:")
for m in motivos:
    print(f"   • {m}")
````

## 🖥️ Seção 8 — Produto de Dados: Simulação do Totem

> Esta seção simula a interface que rodaria no **totem de atendimento ao turista**, gerando previsão + explicação + recomendação de ação em tempo real.

````python
# ============================================================
# PRODUTO DE DADOS — SIMULAÇÃO DO TOTEM
# ============================================================
from IPython.display import display, HTML

def prever_e_explicar(
    data_str: str,
    parque: str,
    temperatura: float,
    precipitacao: float,
    eh_feriado: int = 0,
):
    """
    Função principal do produto de dados.
    Retorna previsão, classificação e explicação SHAP.
    """
    data = pd.to_datetime(data_str)
    parques_lista = df['parque'].unique().tolist()
    parque_enc = parques_lista.index(parque) if parque in parques_lista else 0
    cap_base = int(df[df['parque'] == parque]['capacidade_base'].iloc[0])

    mes   = data.month
    dsem  = data.dayofweek
    eh_fds= int(dsem >= 5)
    saz   = 0.60 * np.sin(2*np.pi*(mes-1)/12) + 0.35*np.sin(2*np.pi*(mes-12)/12+0.5)

    row = {
        'mes_sin':      np.sin(2*np.pi*mes/12),
        'mes_cos':      np.cos(2*np.pi*mes/12),
        'dia_sin':      np.sin(2*np.pi*dsem/7),
        'dia_cos':      np.cos(2*np.pi*dsem/7),
        'eh_fds':       eh_fds,
        'eh_feriado':   eh_feriado,
        'eh_verao':     int(mes in [12,1,2]),
        'eh_ferias_jul':int(mes==7),
        'temperatura':  temperatura,
        'precipitacao': precipitacao,
        'sazonalidade': saz,
        'eh_alta_temp': int(temperatura >= 28),
        'sem_chuva':    int(precipitacao == 0),
        'chuva_forte':  int(precipitacao > 15),
        'fds_x_verao':  eh_fds * int(mes in [12,1,2]),
        'fds_x_feriado':eh_fds * eh_feriado,
        'temp_sem_chuva':temperatura * int(precipitacao == 0),
        'atratividade': np.clip(0.35*saz + 0.25*eh_fds + 0.20*eh_feriado +
                                0.10*(temperatura-15)/20 + 0.10*(1-precipitacao/50), -1, 1),
        'trimestre':    (mes-1)//3 + 1,
        'semana_ano':   data.isocalendar().week,
        'parque_enc':   parque_enc,
    }

    X_in = pd.DataFrame([row])[FEATURES]

    visitantes_prev = int(modelo_final.predict(X_in)[0])
    proba_clf = clf.predict_proba(X_in)[0]
    classe_enc = clf.predict(X_in)[0]
    nivel = le.inverse_transform([classe_enc])[0]

    # SHAP local
    sv = explainer.shap_values(X_in)
    contribs_df = pd.DataFrame({'Feature': FEATURES, 'SHAP': sv[0]})\
                    .sort_values('SHAP', key=abs, ascending=False).head(5)

    # Recomendação operacional
    if nivel == 'Alta':
        cor = '#FF6B6B'
        emoji = '🔴'
        acao = ('ALTA DEMANDA: Ativar equipes extras, abrir filas alternativas, '
                'monitorar capacidade em tempo real e considerar limitação de entrada.')
    elif nivel == 'Média':
        cor = '#FFB400'
        emoji = '🟡'
        acao = ('DEMANDA MODERADA: Operação padrão. Monitorar fluxo de entrada '
                'e manter equipes em alerta para ajuste rápido.')
    else:
        cor = '#00C9A7'
        emoji = '🟢'
        acao = ('BAIXA DEMANDA: Bom momento para manutenção, treinamentos '
                'e atendimento personalizado. Considerar promoções.')

    # Texto SHAP em linguagem natural
    motivos_txt = []
    for _, r in contribs_df.iterrows():
        sinal = '⬆️' if r['SHAP'] > 0 else '⬇️'
        motivos_txt.append(f"{sinal} {r['Feature']}: {r['SHAP']:+.0f} visitantes")

    return visitantes_prev, nivel, cor, emoji, acao, motivos_txt, contribs_df, proba_clf


def exibir_totem(data_str, parque, temperatura, precipitacao, eh_feriado=0):
    vis, nivel, cor, emoji, acao, motivos, df_sh, probas = prever_e_explicar(
        data_str, parque, temperatura, precipitacao, eh_feriado)

    classes = le.classes_
    prob_str = ' | '.join([f"{c}: {p*100:.1f}%" for c, p in zip(classes, probas)])
    motivos_html = ''.join([f"<li style='margin:6px 0;font-size:14px;'>{m}</li>" for m in motivos])

    html = f"""
    <div style="font-family:'Courier New',monospace; max-width:680px; margin:0 auto;
                background:#0f2027; border-radius:16px; overflow:hidden;
                box-shadow:0 8px 40px rgba(0,0,0,0.5);">
      <!-- Header -->
      <div style="background:{cor}22; border-bottom:3px solid {cor};
                  padding:20px 28px; display:flex; justify-content:space-between; align-items:center;">
        <div>
          <div style="font-size:11px;color:#aaa;letter-spacing:2px;margin-bottom:4px;">TOTEM DE ATENDIMENTO · FLEXMEDIA</div>
          <div style="font-size:20px;font-weight:700;color:#fff;">{parque}</div>
          <div style="font-size:12px;color:#aaa;margin-top:4px;">{data_str} · {temperatura}°C · {precipitacao}mm chuva</div>
        </div>
        <div style="text-align:center;">
          <div style="font-size:40px;">{emoji}</div>
          <div style="font-size:11px;color:{cor};font-weight:700;">{nivel.upper()}</div>
        </div>
      </div>
      <!-- Previsão -->
      <div style="padding:20px 28px; border-bottom:1px solid #1e3a4a;">
        <div style="font-size:11px;color:#aaa;letter-spacing:2px;margin-bottom:8px;">PREVISÃO DE VISITANTES</div>
        <div style="font-size:52px;font-weight:700;color:{cor};line-height:1;">{vis:,}</div>
        <div style="font-size:12px;color:#aaa;margin-top:6px;">Probabilidades: {prob_str}</div>
      </div>
      <!-- Ação recomendada -->
      <div style="padding:18px 28px; border-bottom:1px solid #1e3a4a;
                  background:rgba(255,255,255,0.02);">
        <div style="font-size:11px;color:#aaa;letter-spacing:2px;margin-bottom:8px;">RECOMENDAÇÃO OPERACIONAL</div>
        <div style="font-size:13px;color:#e0e0e0;line-height:1.6;">{acao}</div>
      </div>
      <!-- SHAP -->
      <div style="padding:18px 28px;">
        <div style="font-size:11px;color:#aaa;letter-spacing:2px;margin-bottom:12px;">🔍 POR QUÊ ESSA PREVISÃO? (SHAP)</div>
        <ul style="list-style:none;padding:0;margin:0;color:#94d2bd;">{motivos_html}</ul>
      </div>
    </div>
    """
    display(HTML(html))


print("✅ Função do produto definida. Execute os cenários abaixo:")
````

````python
# ============================================================
# CENÁRIO 1 — Feriado no verão (Alta demanda esperada)
# ============================================================
exibir_totem(
    data_str='2026-01-01',
    parque='Chapada dos Veadeiros',
    temperatura=32.0,
    precipitacao=2.5,
    eh_feriado=1
)
````

## ⚠️ Seção 9 — Avaliação Crítica

<div style="background:#1a1a2e; border-radius:12px; padding:28px 32px; font-family:'Courier New',monospace;">

### 🔴 Limitações do Modelo

| # | Limitação | Impacto |
|---|-----------|--------|
| 1 | **Dados sintéticos** — modelo treinado em dados simulados, não histórico real | Alto — métricas otimistas |
| 2 | **Sem séries temporais** — modelos ARIMA/Prophet capturariam melhor autocorrelação | Médio — underfit em longo prazo |
| 3 | **Eventos atípicos** — pandemias, catástrofes naturais não estão modelados | Alto — falha em cenários extremos |
| 4 | **Features climáticas estáticas** — a previsão climática tem incerteza própria | Médio — propagação de erros |
| 5 | **Capacidade real do parque** — limitação física não está no modelo | Alto — pode prever acima da lotação |

### 🟡 Possíveis Melhorias

- **Dados reais do ICMBio** — integrar histórico oficial de visitação
- **Modelo sequencial (LSTM/Prophet)** — capturar dependência temporal
- **Features externas** — Google Trends, preço de passagens, eventos locais
- **Retreinamento automático** — pipeline de MLOps com Airflow + MLflow
- **Intervalo de confiança** — Quantile Regression ou Conformal Prediction
- **Modelo por parque** — especialização ao invés de modelo global

### 🔵 Riscos

- **Overfitting** — R² muito alto pode indicar memorização; validação temporal seria mais robusta
- **Data leakage** — se features futuras vazarem para o treino, métricas serão infladas
- **Viés de seleção** — dados de dias fechados por manutenção podem distorcer a distribuição
- **Deriva de conceito** — padrões de turismo mudam (pós-COVID, novas tendências)
